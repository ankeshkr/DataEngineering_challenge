### Read data

```python
df = spark.read.csv("/dataset/nyc-jobs.csv", header=True,
    inferSchema=True,
    multiLine=True,
    escape='"',
    quote='"',
    ignoreLeadingWhiteSpace=True,
    ignoreTrailingWhiteSpace=True)
df.printSchema()
```

```python
# data cleansing: remove the records where job category is not known(null)
from pyspark.sql.functions import *
df1 = df.filter(col("Job Category").isNotNull())
```

```python
df1.select("Job Category").distinct().count()
```

```python
#Whats the number of jobs posting per category (Top 10)?
```

```python
df1.groupBy("Job Category").agg(count("Job ID").alias("Number of Jobs")).orderBy(desc("Number of Jobs")).show(10)
```

```python
# Whats the salary distribution per job category?
```

```python
from pyspark.sql.functions import col, when, avg, min, max, count

# Calculate estimated salary (average of range)
df_with_avg = df1.withColumn(
    "Estimated Salary",
    (col("Salary Range From") + col("Salary Range To")) / 2
)

# Convert Hourly to Annual (assume 2080 hours/year)
df_converted = df_with_avg.withColumn(
    "Annual Salary",
    when(col("Salary Frequency") == "Hourly", col("Estimated Salary") * 2080)
    .otherwise(col("Estimated Salary"))
)

# Group by Job Category and calculate salary stats
salary_stats = df_converted.groupBy("Job Category").agg(
    count("*").alias("Job Count"),
    avg("Annual Salary").alias("Avg Salary"),
    min("Annual Salary").alias("Min Salary"),
    max("Annual Salary").alias("Max Salary")
).orderBy("Avg Salary", ascending=False)

salary_stats.show()
```

```python
# Whats the job posting having the highest salary per agency?
```

```python
from pyspark.sql.window import Window
```

```python
window = Window.partitionBy("Agency").orderBy(desc("Annual Salary"))
```

```python
# Apply dense_rank and select columns
# df_converted because it contains annual salary
ranked_df = df_converted.withColumn("rank", dense_rank().over(window))
```

```python
ranked_df.select("Agency", "Job Id", "Annual Salary", "rank").where(col("rank") == 1).drop("rank").orderBy(desc("Annual Salary")).show(truncate = False)
```

```python
df_converted.groupBy("Agency").agg(max("Annual Salary").alias("max_salary")).orderBy(desc("max_salary")).show()
```

```python
df_converted.select("Posting Date").show()
```

```python
# Convert posting date to proper date format
#from datetime import datetime
import datetime
df_clean = df.withColumn("Posting_Date", to_date(col("Posting Date"), "MM/dd/yyyy"))
# 2. Filter for last 2 years
current_year = datetime.datetime.now().year
df_recent = df_clean.filter(year(col("Posting_Date")) >= current_year - 2)
#df_recent.printSchema()
# Calculate estimated salary (average of range)
df_with_avg = df.withColumn(
    "Estimated Salary",
    (col("Salary Range From") + col("Salary Range To")) / 2
)

# Convert Hourly to Annual (assume 2080 hours/year)
df_salary = df_with_avg.withColumn(
    "Annual Salary",
    when(col("Salary Frequency") == "Hourly", col("Estimated Salary") * 2080)
    .otherwise(col("Estimated Salary"))
)
# 5. Group by agency and compute average salary
avg_salary_per_agency = df_salary.groupBy("Agency").agg(avg("Annual Salary").alias("Avg_Salary"))

# 6. Show result
avg_salary_per_agency.orderBy(col("Avg_Salary").desc()).show()
```

```python
df_salary.select("Preferred Skills").show()
```

```python
df_skills = df_converted.select("Preferred Skills", "Annual Salary").na.drop()
# 4. Preprocess: lowercase, remove punctuation
df_skills = df_skills.withColumn("Skills_Cleaned",
    regexp_replace(lower(col("Preferred Skills")), "[^a-zA-Z0-9,./+ -]", "")
)

# Tokenize by comma, slash, or semicolon — adjust as needed
df_skills = df_skills.withColumn("Skill", explode(split(col("Skills_Cleaned"), "[,/;+]"))) \
                     .withColumn("Skill", trim(col("Skill")))

# Remove empty strings
df_skills = df_skills.filter(col("Skill") != "")

# Group by skill and calculate average salary
df_skill_salary = df_skills.groupBy("Skill").agg(avg("Annual Salary").alias("Average_Salary"))

# Sort and show top paid skills
df_skill_salary.orderBy(col("Average_Salary").desc()).show(30, truncate=False)
```

### Sample function

```python
def get_salary_frequency(df: DataFrame) -> list:
    row_list = df.select('Salary Frequency').distinct().collect()
    return [row['Salary Frequency'] for row in row_list]
```

### Example of test function

```python
mock_data = [('A', 'Annual'), ('B', 'Daily')]
expected_result = ['Annual', 'Daily']
```

```python
def test_get_salary_frequency(mock_data: list, 
                              expected_result: list,
                              schema: list = ['id', 'Salary Frequency']):  
    mock_df = spark.createDataFrame(data = mock_data, schema = schema)
    assert get_salary_frequency(mock_df) == expected_result
```
